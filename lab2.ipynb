```python
import numpy as np
import math
import scipy.stats as stats
```

```python
full_game = np.array([21824,2768,26344,6644,15104,6652,6492,6568,15488,25260,34176,15164,11344,14824,14400,22136,15168,14384,25764,15144,6608,6640,25904,25636,13496,6840,9684,12476,4848,14744,15172,30184,14856,32632,6624,6664,11260,25356,2904,6652,6628,14548,15004,24220,30192,25844,15068,14576,6452,15156])


short_game = np.array([25844,15144,11428,25832,25984,25728,26192,32284,6792,15060,15040,15244,6552,15164,13984,14432,15264,15052,25796,25984,25752,15160,23736,14336,30428,15348,15476,15072,25160,14572,15060,15168,15352,15684,13672,21976,30380,15076,33556,15328,15028,13684,11572,6540,28832,34244,25752,15784,15024,30356])
print(len(full_game), len(short_game))

# sample mean, standard deviation og margin error til begge spil

meanFull = np.mean(full_game)
stdFull = np.std(full_game, ddof=1)
SE_Full = stdFull/math.sqrt(50)

meanShort = np.mean(short_game)
stdShort = np.std(short_game, ddof=1)
SE_Short = stdShort/math.sqrt(50)

#Confidence interval 95%

CI_Full_high = meanFull + 1.96 * (SE_Full)
CI_Full_low = meanFull - 1.96 * (SE_Full)

print("\nFor the full game:")
print(f'Mean score: {meanFull:.1f}')
print(f'Std: {stdFull :.1f}')
print(f"The confindence interval for full game is [{CI_Full_low:.1f},{CI_Full_high:.1f}]")

CI_Short_high = meanShort + 1.96 * (SE_Short)
CI_Short_low = meanShort - 1.96 * (SE_Short)

print("\nFor the short game:")
print(f'Mean score: {meanShort:.1f}')
print(f'Std: {stdShort :.1f}')
print(f"The confindence interval for short game is [{CI_Short_low:.1f},{CI_Short_high:.1f}]")

# her beregnes degrees of freedom, og så den critical t value
n = 50

varFull = stdFull**2/n
varShort = stdShort**2/n
df= ((varFull+varShort)**2)/(((varFull)**2/(n-1))+((varShort)**2/(n-1)))

t_crit = stats.t.ppf(1 - 0.05/2, df)

tobs = (meanShort - meanFull)/math.sqrt(stdShort**2/50+stdFull**2/50)

print(f"{tobs:.2f}> {t_crit:.2f}. Observed value is higher, therefore can we reject over nulhypotesen. And accept the alternative")


```
